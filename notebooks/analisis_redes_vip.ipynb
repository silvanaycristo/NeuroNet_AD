```python
from pathlib import Path
import pandas as pd
import networkx as nx

ruta = Path("/home/rafadiaz/export/NeuroNet_AD/aracne_vip")

condiciones = ["Vip_High", "Vip_Intermediate", "Vip_Low", "Vip_Not_AD"]

for cond in condiciones:
    tsv = ruta / f"{cond}_hvg.tsv" / "nobootstrap_network.txt"
    out = ruta / f"{cond}_network.graphml"
    
    df = pd.read_csv(tsv, sep="\t")
    G = nx.from_pandas_edgelist(df, source="Regulator", target="Target", edge_attr="MI", create_using=nx.DiGraph())
    nx.write_graphml(G, out)
    
    print(f"{cond}: {G.number_of_nodes()} nodos, {G.number_of_edges()} aristas → {out.name}")
```
